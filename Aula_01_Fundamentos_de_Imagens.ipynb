```python
# Representa e manipula matrizes
import numpy as np
# Exibe as imagens
import matplotlib.pyplot as plt

# Ler, converter, e redimensionar imagens
from PIL import Image
# Acessa imagens de exemplo
from skimage import data, color
```

```python
def mostrar_imagem(imagem, titulo, cmap=None, barra_cores=False, tamanho=(4,4)):
    # Definindo Tamanho da imagem
    plt.figure(figsize=tamanho);

    plt.imshow(imagem, cmap=cmap, vmin=0, vmax=255)
    plt.title(titulo)
    plt.axis("off")

    if barra_cores:
        plt.colorbar()

    plt.show()
```

```python
# Monando minha imagem
imagem_pequena = np.array([
    [0, 40, 80, 120],
    [40, 80, 120, 160],
    [80, 120, 160, 200],
    [120, 160, 200, 255],
], dtype=np.uint8)

print(imagem_pequena)
```

```python
# Tamanho da imagem plotada
plt.figure(figsize=(6,6))
plt.imshow(imagem_pequena, cmap='gray', vmin=0, vmax=255)
# Definindo Título
plt.title("Imagem Teste")
#Retirar Eixos
plt.axis(False)
# Imprimindo com Colorbar
plt.colorbar(label='Intensidade')
# Adicionando Grade
plt.grid(color="red", linewidth=1)
# Mostrando imagem na tela
plt.show()
```

```python
# Tamanho da imagem plotada
plt.figure(figsize=(6,6))
plt.imshow(imagem_pequena, cmap='gray', vmin=0, vmax=255)

# Para cada linha dento do range de (numero máximo de linhas)
for linha in range(imagem_pequena.shape[0]):
    # Para cada coluna dentro do range de (numero máximo de colunas)
    for coluna in range(imagem_pequena.shape[1]):
        # Pegando valor daquela posição
        valor = imagem_pequena[linha, coluna]

        cor_texto = "white" if valor < 128 else "black"

        plt.text(
            coluna,
            linha,
            str(valor),
            ha= "center",
            va= "center",
            color= cor_texto,
            fontsize=12
        )

plt.title('Intensidade de cada pixel')

plt.xticks(range(4))
plt.yticks(range(4))

plt.show()
```

```python
linha = 2
coluna = 3
# Pegando valor de um único pixel
valor = imagem_pequena[linha, coluna]

print(f"Pixel na linha {linha} e coluna {coluna}: {valor}")
```

```python
# Criando uma imagem cópia
imagem_pixel_alterado = imagem_pequena.copy()

imagem_pixel_alterado[1,2] = 255

print(imagem_pequena)
print(imagem_pixel_alterado)
```

```python
# Criando imagem com 1 linha e 2 colunas e definindo tamanho dela
fig, eixos = plt.subplots(1, 2, figsize=(10,4))

# No primeiro eixo
eixos[0].imshow(imagem_pequena, cmap="gray", vmin=0, vmax=255)
eixos[0].set_title("Original")
eixos[0].axis("off")

# No segundo eixo
eixos[1].imshow(imagem_pixel_alterado, cmap="gray", vmin=0, vmax=255)
eixos[1].set_title("Pixel Alterado")
eixos[1].axis("off")
```

```python
# Propriedades de uma imagem
print("Shape: ", imagem_pequena.shape)
print("Numero de Dimensões: ", imagem_pequena.ndim)
print("Tipo de Dado: ", imagem_pequena.dtype)
print("Menor Intensidade: ", imagem_pequena.min())
print("Maior Intensidade: ", imagem_pequena.max())
print("Média de Intensidade: ", imagem_pequena.mean())
```

```python
# Adicionando valor '0' a todas a posições da matriz
imagem_preta = np.zeros((4,4), dtype=np.uint8)

print(imagem_preta)

mostrar_imagem (
    imagem_preta,
    titulo="Imagem toda Preta",
    cmap="gray"
) 
```

```python
# Adicionando valor específico a todas a posições da matriz
imagem_branca = np.full((4,4), 120, dtype=np.uint8)

print(imagem_branca)

mostrar_imagem (
    imagem_branca,
    titulo="Imagem toda Branca",
    cmap="gray"
) 
```

```python
altura = 200

# Cria um intervalo de valores com base em dois extremos
linha_gradiente = np.linspace(0, 255, dtype=np.uint8)

# Cria matrix, repetindo valores (definidos no primeiro parâmetro) ao longo das linhas e colunas
gradiente_horizontal = np.tile(linha_gradiente, (altura, 1))

mostrar_imagem(
    gradiente_horizontal,
    titulo= "Gradiente Horizontal",
    cmap="gray",
    barra_cores=True
)

print(linha_gradiente)
```

```python
largura = 510
# Cria um intervalo de valores com base em dois extremos
coluna_gradiente = np.linspace(0, 255, dtype=np.uint8)

print(coluna_gradiente)

gradiente_vertical = np.tile(coluna_gradiente[:, np.newaxis], (1, largura))

mostrar_imagem(
    gradiente_vertical,
    titulo= "Gradiente Vertical",
    cmap="gray",
    barra_cores=True
)
```

```python
imagem_retangulo = np.zeros((300,400), dtype=np.uint8)

imagem_retangulo[80:220, 120:300] = 255

mostrar_imagem(
    imagem_retangulo,
    titulo="Imagem Retângulo",
    cmap="gray"
)
```

```python
imagem_bloco = np.zeros((300,500), dtype=np.uint8)

# ':' = Pega todos os valores de um intervalo definido, caso não defina pega todos os valores
imagem_bloco[:, 0:100] = 0
imagem_bloco[:, 100:200] = 64
imagem_bloco[:, 200:300] = 128
imagem_bloco[:, 300:400] = 192
imagem_bloco[:, 400:500] = 250

mostrar_imagem(
    imagem_bloco,
    titulo= "Faixas de Cor",
    cmap="gray",
    barra_cores=True
)
```

```python
imagem_formas = np.zeros((400,600), dtype=np.uint8)

imagem_formas[50:180, 55:220] = 255

imagem_formas[230:350, 80:200] = 255

y, x = np.indices(imagem_formas.shape)
mascara = (x - 430) ** 2 + (y - 200) ** 2 <= 90 ** 2
imagem_formas[mascara] = 255

mostrar_imagem(
    imagem_formas,
    titulo="Imagem com Formas",
    cmap="gray",
    barra_cores=True
)
```

```python
# Criando um padrão quadriculado
# Imagem 256 pixels / 256 pixels
tamanho = 256
# Tamanho dos blocos de 32 pixels / 32 pixels
bloco = 32


linha, coluna = np.indices((tamanho, tamanho))

tabuleiro = (
    ((linha // bloco) + (coluna // bloco)) % 2
    * 255
).astype(np.uint8)

mostrar_imagem(
    tabuleiro,
    titulo= "Padrão Quadriculado",
    cmap="gray"
)
```

```python
canal_vermelho = imagem_rgb[:, :, 0] 
canal_verde = imagem_rgb[:, :, 1] 
canal_azul = imagem_rgb[:, :, 2] 

fig, eixos = plt.subplots(1, 3, figsize=(18,6))

eixos[0].imshow(canal_vermelho, vmin=0, vmax=255)
eixos[0].set_title("Canal Vermelho")
eixos[0].axis("off")

eixos[1].imshow(canal_verde, vmin=0, vmax=255)
eixos[1].set_title("Canal Verde")
eixos[1].axis("off")

eixos[2].imshow(canal_azul, vmin=0, vmax=255)
eixos[2].set_title("Canal Azul")
eixos[2].axis("off")

```

```python
imagem_float = imagem_rgb

cinza_media = imagem_float.mean(axis=2)
mostrar_imagem(
    cinza_media,
    titulo="Imagem Cinza",
    cmap="gray"
)
```

```python
cinza_ponderado = (
    0.299 * imagem_float[:, :, 0]+
    0.587 * imagem_float[:, :, 1]+
    0.114 * imagem_float[:, :, 2]
)

mostrar_imagem(
    cinza_ponderado,
    titulo="Tons de Cinza (Ponderado)",
    cmap="gray"
)
```

```python
cinza_skimage = color.rgb2gray(imagem_rgb)

print(cinza_skimage.dtype)
print(cinza_skimage.min())
print(cinza_skimage.max())

mostrar_imagem(
    cinza_skimage,
    titulo="Tons de Cinza (Skimage)",
    cmap="gray"
)
```

```python
recorte = imagem_rgb[80:200, 150:370]

mostrar_imagem(
    recorte,
    titulo="Recorte Imagem"
)
```

```python
imagem_com_marcacao = imagem_rgb.copy()

linha_inicial = 80
linha_final = 280
coluna_inicial = 150
coluna_final = 370
espessura = 1

# Linha Superior
imagem_com_marcacao[
    linha_inicial:linha_inicial + espessura,
    coluna_inicial:coluna_final,
] = [255, 0, 0]

# Linha Inferior
imagem_com_marcacao[
    linha_final - espessura:linha_final,
    coluna_inicial:coluna_final,
] = [255, 0, 0]

# Linha Esquerda
imagem_com_marcacao[
    linha_inicial:linha_final,
    coluna_inicial:coluna_inicial + espessura,
] = [255, 0, 0]

# Linha Direita
imagem_com_marcacao[
    linha_inicial:linha_final,
    coluna_final - espessura:coluna_final,
] = [255, 0, 0]


mostrar_imagem(
    imagem_com_marcacao,
    titulo="Imagem Marcada"
)




```

```python
imagem_pixelada = imagem_rgb.copy()

regiao = imagem_pixelada[80:280, 150:370]

regiao_pil = Image.fromarray(regiao)

regiao_reduzida = regiao_pil.resize((20, 20), Image.Resampling.NEAREST)

regiao_ampliada = regiao_reduzida.resize(
    (regiao.shape[1], regiao.shape[0]),
    Image.Resampling.NEAREST
)

imagem_pixelada[80:280, 150:370] = np.array(regiao_ampliada)

mostrar_imagem(
    imagem_pixelada,
    titulo="Imagem Pixelada"
)
```

```python
# Espelhamento Vertical
# Para todas as colunas traga as linhas de trás para frente
imagem_invertida_vertical = imagem_rgb[::-1, :]

mostrar_imagem(
    imagem_invertida_vertical,
    titulo="Imagem Invetida Verticalmente"
)
```

```python
# Espelhamento Horizontal
# Para todas as linhas traga as colunas de trás para frente
imagem_invertida_horizontal = imagem_rgb[:, ::-1]

mostrar_imagem(
    imagem_invertida_horizontal,
    titulo="Imagem Invetida Horizontalmente"
)
```

```python
rotacao90 = np.rot90(imagem_rgb, k=1) 
rotacao180 = np.rot90(imagem_rgb, k=2) 
rotacao270 = np.rot90(imagem_rgb, k=3) 

fig, eixos = plt.subplots(1, 3, figsize=(18,6))

eixos[0].imshow(rotacao90, vmin=0, vmax=255)
eixos[0].set_title("Cinza Médio")
eixos[0].axis("off")

eixos[1].imshow(rotacao180, vmin=0, vmax=255)
eixos[1].set_title("Cinza Ponderado")
eixos[1].axis("off")

eixos[2].imshow(rotacao270, vmin=0, vmax=255)
eixos[2].set_title("Cinza Skimage")
eixos[2].axis("off")
```

```python
# Aumentando o Brilho
# Garantindo não estouro dos pixels
imagem_cinza_16bits = cinza_ponderado.astype(np.uint16)

# Para cada valor, aplica filtro convertendo valores para dentro do intervalo 
imagem_clara = np.clip(
    imagem_cinza_16bits + 50,
    0,
    255
).astype(np.uint8)

mostrar_imagem(
    imagem_clara,
    titulo="Teste de Imagem",
    cmap="gray"
)
```

```python
imagem_escura = np.clip(
    imagem_cinza_16bits - 50,
    0,
    255
).astype(np.uint8)

mostrar_imagem(
    imagem_escura,
    titulo="Teste de Imagem",
    cmap="gray"
)
```

```python
# Imagem Negativa
negativo = 255 - cinza_ponderado

mostrar_imagem(
    negativo,
    titulo="Negativo da Imagem",
    cmap="gray"
)
```
